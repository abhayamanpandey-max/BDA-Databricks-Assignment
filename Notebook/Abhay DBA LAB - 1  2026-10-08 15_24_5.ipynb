```python
# Load CSV into DataFrame 
df = spark.read.csv("/Volumes/abhay_data/default/abhay_bda_data/student_scores.csv", header=True, inferSchema=True) 
 
# Show first 5 rows 
df.show(5) 
```

```python
# Register DataFrame as SQL view 
df.createOrReplaceTempView("students") 
```

```python
%sql
-- Select all records 
SELECT * FROM students; 
```

```python
# Count of the dataset
df.count()
```

```python
%sql
-- Count students by gender 
SELECT gender, COUNT(*) AS total_students 
FROM students 
GROUP BY gender; 
```

```python
%sql
-- Average math score by parental education 
SELECT parental_level_of_education, AVG(math_score) AS avg_math 
FROM students 
GROUP BY parental_level_of_education 
ORDER BY avg_math DESC; 
```

```python
%sql
-- Top 5 students by reading score
SELECT student_id, reading_score 
FROM students 
ORDER BY reading_score DESC 
LIMIT 5;
```
